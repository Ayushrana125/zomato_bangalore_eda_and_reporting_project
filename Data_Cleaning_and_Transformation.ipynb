```python
# Understanding the Reviews from the outlets table

outlets["reviews_list"]
pd.set_option("display.max_colwidth", None)
outlets["reviews_list"].head(1)
```

```python
# Working on Reviews from Outlets table and creating a seperate dimension table for it

import ast
import pandas as pd

ratings_work = (
    outlets[["outlet_id", "restaurant_id", "reviews_list"]]
    .dropna(subset=["reviews_list"])
    .copy()
)
```

```python
def parse_reviews(review_text):
    try:
        reviews = ast.literal_eval(review_text)

        if isinstance(reviews, list):
            return reviews
        
        return []
    except:
        return []
```

```python
ratings_work["parsed_reviews"] = ratings_work["reviews_list"].apply(parse_reviews)
```

```python
#Seperating each review as a row

ratings = (
    ratings_work[["outlet_id", "restaurant_id", "parsed_reviews"]]
    .explode("parsed_reviews")
    .dropna(subset=["parsed_reviews"])
    .copy()
)
```

```python
# Splitting tuple into rate and feedback

ratings["rate"] = ratings["parsed_reviews"].apply(lambda x: x[0] if isinstance(x, tuple) and len(x) > 0 else None)
ratings["feedback"] = ratings["parsed_reviews"].apply(lambda x: x[1] if isinstance(x, tuple) and len(x) > 1 else None)
```

```python
# Cleaning reviews and rate

ratings["rate"] = (
    ratings["rate"]
    .str.replace("Rated", "", regex=False)
    .str.strip()
)

ratings["rate"] = pd.to_numeric(ratings["rate"], errors="coerce")
```

```python
# Cleaning feedback text 

ratings["feedback"] = (
    ratings["feedback"]
    .str.replace("RATED\n", "", regex=False)
    .str.strip()
)
```

```python
# Adding rating_id

ratings = ratings.drop(columns=["parsed_reviews"])

ratings = ratings.reset_index(drop=True)

ratings["rating_id"] = ["RA" + str(i) for i in range(1, len(ratings) + 1)]

ratings = ratings[["rating_id", "outlet_id", "restaurant_id", "rate", "feedback"]]
```

```python
import pandas as pd
outlets = pd.read_csv("D:\-- Ayush Rana --\-- Data Analysis --\Data Analyst Projects\Zomato Bangalore Restaurants\Tables_Created\outlets.csv")
```

```python
# Adding Longitude & Latitude for locations and a seperate table

from geopy.geocoders import Nominatim
from geopy.extra.rate_limiter import RateLimiter
import pandas as pd

locations = (
    outlets[["location"]]
    .dropna()
    .drop_duplicates()
    .reset_index(drop=True)
)

locations["location_query"] = locations["location"] + ", Bengaluru, Karnataka, India"
```

```python
geolocator = Nominatim(user_agent="zomato_bangalore_eda_project")

geocode = RateLimiter(
    geolocator.geocode,
    min_delay_seconds=1
)

locations["geo"] = locations["location_query"].apply(geocode)
```

```python
# Extracting lattitude and longitude from above results

locations["latitude"] = locations["geo"].apply(lambda x: x.latitude if x else None)
locations["longitude"] = locations["geo"].apply(lambda x: x.longitude if x else None)
```

```python
locations = locations.drop(columns=["geo"])
```

```python
outlets = outlets.merge(
    locations[["location", "latitude", "longitude"]],
    on="location",
    how="left"
)
```

```python
outlets.to_csv("Tables_Created/outlets.csv", index=False)
locations.to_csv("Tables_Created/locations.csv", index=False)
```
